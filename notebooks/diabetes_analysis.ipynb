# Diabetes Risk Analysis – Exploratory Data Analysis

This notebook provides a walkthrough of the dataset and key analytical findings
that power the Streamlit dashboard.

**Dataset:** Synthetic Pima Indians Diabetes Dataset (768 patients)  
**Columns:** Pregnancies, Glucose, BloodPressure, SkinThickness, Insulin, BMI,
DiabetesPedigreeFunction, Age, Outcome


```python
import sys
import os
sys.path.insert(0, os.path.join(os.getcwd(), '..'))

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

from analysis.data_loader import load_data
from analysis.metrics import (
    compute_key_metrics,
    compute_risk_distribution,
    compute_bmi_distribution,
)
from analysis.visualizations import (
    correlation_heatmap,
    glucose_age_scatter,
    risk_pie_chart,
    feature_importance_chart,
)

df = load_data()
print(f'Dataset shape: {df.shape}')
df.head()
```

```python
metrics = compute_key_metrics(df)
for k, v in metrics.items():
    print(f'{k:30s}: {v}')
```

```python
risk_df = compute_risk_distribution(df)
print(risk_df)
risk_pie_chart(risk_df).show()
```

## 5. BMI Distribution

```python
bmi_df = compute_bmi_distribution(df)
print(bmi_df)

fig = px.bar(
    bmi_df,
    x='BMICategory',
    y='Count',
    title='BMI Category Distribution',
    template='plotly_dark',
)
fig.show()
```

## 8. Key Findings

| Insight | Detail |
|---|---|
| Top predictor | Glucose is the single strongest predictor of diabetes |
| BMI impact | Obese patients (BMI ≥ 30) have significantly higher diabetes rates |
| Age trend | Average glucose rises with age, especially in the 46–60 group |
| Risk distribution | ~32–40% of patients fall into the High-risk category |
| BP & risk | High-risk patients show slightly elevated blood pressure on average |
